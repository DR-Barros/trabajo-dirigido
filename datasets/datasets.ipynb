```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs, make_moons, make_circles
from sklearn.decomposition import PCA

# Estilo de gráficos
plt.style.use("seaborn-v0_8")

# Dataset 1: Blobs (distribución gaussiana)
X_blobs, y_blobs = make_blobs(n_samples=300, centers=2, cluster_std=1.0, random_state=42)

# Dataset 2: Moons (forma de luna)
X_moons, y_moons = make_moons(n_samples=300, noise=0.1, random_state=42)

# Dataset 3: Circles (forma circular)
X_circles, y_circles = make_circles(n_samples=300, noise=0.05, factor=0.2, random_state=42)

# Dataset 4: Blobs con más características
X, y = make_blobs(n_samples=500, centers=4, n_features=5, random_state=42)

# Visualización
fig, axs = plt.subplots(1, 4, figsize=(20, 4))

# Blobs
axs[0].scatter(X_blobs[:, 0], X_blobs[:, 1], c=y_blobs, cmap='viridis')
axs[0].set_title("make_blobs (clusters gaussianos)")

# Moons
axs[1].scatter(X_moons[:, 0], X_moons[:, 1], c=y_moons, cmap='viridis')
axs[1].set_title("make_moons (no lineal)")

# Circles
axs[2].scatter(X_circles[:, 0], X_circles[:, 1], c=y_circles, cmap='viridis')
axs[2].set_title("make_circles (concéntricos)")

# Blobs con más características
pca = PCA(n_components=2)
X_reduced = pca.fit_transform(X)
axs[3].scatter(X_reduced[:, 0], X_reduced[:, 1], c=y, cmap='viridis')
axs[3].set_title("make_blobs (5 características, PCA)")

plt.tight_layout()
plt.show()
```

```python
#agregale columnas con ruido
import numpy as np

def add_noise(X, noise_level=0.1):
    noise = np.random.normal(0, noise_level, X.shape)
    return X + noise
# Agregar ruido a los datasets
X_blobs_noisy = add_noise(X_blobs, noise_level=0.5)
X_moons_noisy = add_noise(X_moons, noise_level=0.1)
X_circles_noisy = add_noise(X_circles, noise_level=0.05)
X_blobs_noisy_5d = add_noise(X, noise_level=1)
# Visualización con ruido
fig, axs = plt.subplots(1, 4, figsize=(20, 4))
# Blobs con ruido
X_blobs_pca = PCA(n_components=2).fit_transform(X_blobs_noisy)
axs[0].scatter(X_blobs_pca[:, 0], X_blobs_pca[:, 1], c=y_blobs, cmap='viridis')
axs[0].set_title("make_blobs con ruido (PCA)")
# Moons con ruido
X_moons_pca = PCA(n_components=2).fit_transform(X_moons_noisy)
axs[1].scatter(X_moons_pca[:, 0], X_moons_pca[:, 1], c=y_moons, cmap='viridis')
axs[1].set_title("make_moons con ruido (PCA)")
# Circles con ruido
X_circles_pca = PCA(n_components=2).fit_transform(X_circles_noisy) 
axs[2].scatter(X_circles_pca[:, 0], X_circles_pca[:, 1], c=y_circles, cmap='viridis')
axs[2].set_title("make_circles con ruido (PCA)")

# Blobs con más características y ruido
X_blobs_noisy_5d_pca = PCA(n_components=2).fit_transform(X_blobs_noisy_5d)
axs[3].scatter(X_blobs_noisy_5d_pca[:, 0], X_blobs_noisy_5d_pca[:, 1], c=y, cmap='viridis')
axs[3].set_title("make_blobs 5D con ruido (PCA)")
plt.tight_layout()
plt.show()
```
